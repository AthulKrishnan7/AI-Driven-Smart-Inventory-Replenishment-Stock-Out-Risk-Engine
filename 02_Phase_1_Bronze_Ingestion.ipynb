```python
from pyspark.sql import Row
```

```python
spark.sql("USE inventory_ai")

sales_raw_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv("/Volumes/inventory/default/train/train.csv")

sales_raw_df.show(5)
sales_raw_df.printSchema()
```

```python
sales_raw_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("inventory_ai.bronze_sales_raw")
```

## Inventory Snapshot

```python
from pyspark.sql import Row

inventory_data = [
    Row(item=1, stock_on_hand=120),
    Row(item=2, stock_on_hand=80),
    Row(item=3, stock_on_hand=60)
]

inventory_raw_df = spark.createDataFrame(inventory_data)

inventory_raw_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("inventory_ai.bronze_inventory_raw")
```

Supplier lead time

```python
supplier_data = [
    Row(item=1, supplier_id="S1", lead_time_days=5),
    Row(item=2, supplier_id="S2", lead_time_days=7),
    Row(item=3, supplier_id="S1", lead_time_days=10)
]

supplier_raw_df = spark.createDataFrame(supplier_data)

supplier_raw_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("inventory_ai.bronze_suppliers_raw")
```
